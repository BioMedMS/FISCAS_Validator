```python
%matplotlib widget
def convert_to_float_and_round(value, decimal=4):
    print(value)
    if isinstance(value, (list, tuple)):
        return [convert_to_float_and_round(item) for item in value]
    elif isinstance(value, np.ndarray):
        return np.array([convert_to_float_and_round(item) for item in value])
    else:
        return round(float(value), decimal)
```
